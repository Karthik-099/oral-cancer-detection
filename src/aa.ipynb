```python
import pandas as pd
pd.options.display.max_rows = 1000
pd.options.display.max_columns = 1000

df = pd.read_csv(r"C:\Users\a\Desktop\Oral\data\dataset.csv")

# Drop rows with NaNs in key columns
df.dropna(subset=['Risk', 'Specialist diagnosis'], inplace=True)

# Replace '-' with 0
df.replace('-', 0, inplace=True)

df['FLW username'] = df['FLW username'].apply(lambda x: int(str(x).split()[1]))


# Remove rows where both Risk and Specialist diagnosis are 0
df = df[(df['Risk'] != 0) | (df['Specialist diagnosis'] != 0)]

# Map tobacco/alcohol use: 'never'->0, 'past'->1, 'current'->2
use_map = {'never': 0, 'past': 1, 'current': 2}
for col in ['Cigarette/Bidi', 'Smokeless tobacco', 'Areca nut', 'Alcohol']:
    df[col] = df[col].astype(str).str.lower().map(use_map).fillna(0)

# Encode gender: 'male'->1, 'female'->0
df['Gender'] = df['Gender'].astype(str).str.lower().map({'male': 1, 'female': 0})

# Encode study setting: 'hospital'->1, 'primary care'->0
df['Study setting'] = df['Study setting'].astype(str).str.lower().map({'hospital': 1, 'primary care': 0})

# Encode Risk: 'low risk'->0, 'high risk'->1
df['Risk'] = df['Risk'].astype(str).str.lower().map({'low risk': 0, 'high risk': 1})

# Encode Specialist diagnosis: 'non suspicious'->0, 'suspicious'->1
df['Specialist diagnosis'] = df['Specialist diagnosis'].astype(str).str.lower().map({
    'non suspicious': 0,
    'non-suspicious': 0,
    'suspicious': 1
})

# Encode 'yes' -> 1 and 'no' -> 0 for these two binary columns
df['Able to open mouth'] = df['Able to open mouth'].astype(str).str.lower().map({'yes': 1, 'no': 0}).fillna(0)
df['Lesion/patch'] = df['Lesion/patch'].astype(str).str.lower().map({'yes': 1, 'no': 0}).fillna(0)
```

```python
# the correlation between 'Age' and 'Risk'
correlation = df['Age'].corr(df['Risk'])

print(f"The correlation between Age and Risk is: {correlation:.2f}")
```

```python
habits = ['Cigarette/Bidi', 'Smokeless tobacco', 'Areca nut', 'Alcohol']

print("Correlation of habits with Risk:")
for habit in habits:
    correlation = df[habit].corr(df['Risk'])
    print(f"Correlation between '{habit}' and Risk: {correlation:.2f}")
```

```python
#Between  Specialist diagnosis and Risk
diseases = ['Oral cancer', 'Oral leukoplakia', 'Oral lichen planus',
            'Oral submucous fibrosis', 'Oral candidiasis', 'Oral hairy leukoplakia',
            'Oral dysplasia', 'Oral erythroplakia']

print("Correlation of disease with Risk:")
for disease_type in diseases:
    if disease_type in df.columns:
        correlation = df[disease_type].corr(df['Risk'])
        print(f"Correlation between '{disease_type}' and Risk: {correlation:.2f}")
    else:
        print(f"Column '{disease_type}' not found in DataFrame.")
```
